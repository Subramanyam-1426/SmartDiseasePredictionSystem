```python
import pandas as pd
from sklearn.preprocessing import LabelEncoderfrom sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
# Load training dataset
train_data = pd.read_csv(
    r"C:\Users\nanik\OneDrive\Desktop\ML external\archive\Training.csv"
)

# Load testing dataset
test_data = pd.read_csv(
    r"C:\Users\nanik\OneDrive\Desktop\ML external\archive\Testing.csv"
)

print("Training data loaded successfully")
print("Testing data loaded successfully")
```

```python
print(train_data)
```

```python
print(train_data.head())
```

```python
print(train_data.tail())
```

```python
print("Training data shape:", train_data.shape)
print("Testing data shape:", test_data.shape)
```

```python
xtrain = train_data.drop(columns=["prognosis", "Unnamed: 133"])
ytrain = train_data["prognosis"]

xtest = test_data.drop(columns=["prognosis"])
ytest = test_data["prognosis"]
```

```python
print("Training Features:", xtrain.shape)
print("Training Target:", ytrain.shape)

print("Testing Features:", xtest.shape)
print("Testing Target:", ytest.shape)
```

```python
print("Number of Features:", len(xtrain.columns))
print("Feature Names:")
print(xtrain.columns.tolist())
```

```python
encoder = LabelEncoder()

ytrain = encoder.fit_transform(ytrain)
ytest = encoder.transform(ytest)
```

```python
print("Disease Classes:")
print(encoder.classes_)

print("Number of Disease Classes:", len(encoder.classes_))
```

```python
model = KNeighborsClassifier(
    n_neighbors=1,
    metric='euclidean'
)
```

```python
model.fit(xtrain, ytrain)

print("Model trained successfully!")
```

```python
ypred = model.predict(xtest)

print("Predicted classes:", ypred)
```

```python
accuracy = accuracy_score(ytest, ypred) * 100

print("Accuracy:", accuracy, "%")
```

```python
disease = encoder.inverse_transform(prediction)

print("Predicted Disease:", disease[0])
```
