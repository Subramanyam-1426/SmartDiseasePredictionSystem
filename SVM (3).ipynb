```python
# Import pandas for handling the dataset
import pandas as pd
```

```python
# Load the training dataset
train_data = pd.read_csv(
    r"Training.csv"
)

print("Training data loaded successfully")
```

```python
print(train_data)
```

```python
# Display the first five rows of the training dataset
print(train_data.head())
```

```python
# Load the testing dataset
test_data = pd.read_csv(
    r"Testing.csv"
)

print("Testing data loaded successfully")
```

```python
# Display the first five rows of the training dataset
print(train_data.tail())
```

```python
# Display the number of rows and columns in both datasets
print("Training data shape:", train_data.shape)
print("Testing data shape:", test_data.shape)
```

```python
# Remove the target column and unnecessary column from training data
xtrain = train_data.drop(
    columns=["prognosis", "Unnamed: 133"]
)

print("Training features separated successfully")
```

```python
# Select the disease column as the training target
ytrain = train_data["prognosis"]

print("Training target separated successfully")
```

```python
# Remove the target column from testing data
xtest = test_data.drop(columns=["prognosis"])

print("Testing features separated successfully")
```

```python
# Select the disease column as the testing target
ytest = test_data["prognosis"]

print("Testing target separated successfully")
```

```python
# Import LabelEncoder to convert disease names into numbers
from sklearn.preprocessing import LabelEncoder
```

```python
# Create an object of LabelEncoder
encoder = LabelEncoder()

# Display confirmation
print("Label encoder created successfully")
```

```python
# Convert disease names into numerical labels
ytrain = encoder.fit_transform(ytrain)
ytest = encoder.transform(ytest)

print("Disease names encoded successfully")
```

```python
# Import the Support Vector Machine classifier
from sklearn.svm import SVC

print("SVM classifier imported successfully")
```

```python
# Create the SVM model using a linear kernel
model = SVC(
    kernel="linear",
    random_state=1
)

print("SVM model created successfully")
```

```python
# Train the SVM model using the training data
model.fit(xtrain, ytrain)

print("Model trained successfully")
```

```python
# Predict disease labels for the testing data
ypred = model.predict(xtest)

# Display confirmation
print("Prediction completed successfully")
```

```python
# Import accuracy_score to evaluate the model
from sklearn.metrics import accuracy_score

# Display confirmation
print("Accuracy metric imported successfully")
```

```python
# Calculate the prediction accuracy of the SVM model
accuracy = accuracy_score(ytest, ypred) * 100

# Display the accuracy
print("Accuracy:", accuracy, "%")
```

```python
new_patient = pd.DataFrame(
    0,
    index=[0],
    columns=xtrain.columns
)

print(new_patient)
```

```python
for symptom in symptoms:
    if symptom in new_patient.columns:
        new_patient[symptom] = 1
    else:
        print("Symptom not found:", symptom)

print(new_patient)
```

```python
prediction = model.predict(new_patient)

print("Predicted Class:", prediction)
```

```python
disease = encoder.inverse_transform(prediction)

print("Predicted Disease:", disease[0])
```
