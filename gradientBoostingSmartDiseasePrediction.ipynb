```python
# Cell 1: Import the libraries required for data handling, label encoding, Gradient Boosting, and accuracy calculation.
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score
```

```python
# Cell 2: Load the training and testing datasets.
# Change these paths if your CSV files are stored in a different folder.
train_data = pd.read_csv(r"Training.csv")
test_data = pd.read_csv(r"Testing.csv")

print("Training data loaded successfully")
print("Testing data loaded successfully")
```

```python
# Cell 3: Display the complete training dataset.
print(train_data)
```

```python
# Cell 4: Display the first five rows to understand the structure of the dataset.
print(train_data.head())
```

```python
# Cell 5: Display the last five rows of the training dataset.
print(train_data.tail())
```

```python
# Cell 6: Check the number of rows and columns in the training and testing datasets.
print("Training data shape:", train_data.shape)
print("Testing data shape:", test_data.shape)
```

```python
# Cell 7: Separate the symptom features from the disease target.
# 'prognosis' is the target column, while the remaining columns represent symptoms.
xtrain = train_data.drop(columns=["prognosis", "Unnamed: 133"], errors="ignore")
ytrain = train_data["prognosis"]

xtest = test_data.drop(columns=["prognosis", "Unnamed: 133"], errors="ignore")
ytest = test_data["prognosis"]
```

```python
# Cell 8: Display the dimensions of training and testing features and targets.
print("Training Features:", xtrain.shape)
print("Training Target:", ytrain.shape)

print("Testing Features:", xtest.shape)
print("Testing Target:", ytest.shape)
```

```python
# Cell 9: Display the number and names of symptom features used by the model.
print("Number of Features:", len(xtrain.columns))
print("Feature Names:")
print(xtrain.columns.tolist())
```

```python
# Cell 10: Convert disease names into numerical labels so the Gradient Boosting classifier can process them.
encoder = LabelEncoder()

ytrain = encoder.fit_transform(ytrain)
ytest = encoder.transform(ytest)
```

```python
# Cell 11: Display the disease classes and the total number of disease classes.
print("Disease Classes:")
print(encoder.classes_)

print("Number of Disease Classes:", len(encoder.classes_))
```

```python
# Cell 12: Create the Gradient Boosting classifier.
# n_estimators = number of decision trees, learning_rate controls each tree's contribution,
# and max_depth controls the depth of each weak learner.
model = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

print(model)
```

```python
# Cell 13: Train the Gradient Boosting model using the training symptoms and disease labels.
model.fit(xtrain, ytrain)

print("Gradient Boosting model trained successfully!")
```

```python
# Cell 14: Predict disease classes for the testing dataset.
ypred = model.predict(xtest)

print("Predicted classes:", ypred)
```

```python
# Cell 15: Calculate and display the prediction accuracy on the testing dataset.
accuracy = accuracy_score(ytest, ypred) * 100

print("Accuracy:", accuracy, "%")
```

```python
# Cell 16: Ask the user to enter symptoms for a new patient.
symptoms = input("Enter the symptoms separated by comma: ")

symptoms = [s.strip() for s in symptoms.split(",")]

print("Entered Symptoms:", symptoms)
```

```python
# Cell 17: Create a new patient record with all symptoms initially set to zero.
new_patient = pd.DataFrame(
    0,
    index=[0],
    columns=xtrain.columns
)

print(new_patient)
```

```python
# Cell 18: Mark the symptoms entered by the user as 1 in the new patient record.
for symptom in symptoms:
    if symptom in new_patient.columns:
        new_patient[symptom] = 1
    else:
        print("Symptom not found:", symptom)

print(new_patient)
```

```python
# Cell 19: Use the trained Gradient Boosting model to predict the disease class for the new patient.
prediction = model.predict(new_patient)

print("Predicted Class:", prediction)
```

```python
# Cell 20: Convert the predicted numerical class back into the original disease name.
disease = encoder.inverse_transform(prediction)

print("Predicted Disease:", disease[0])
```

```python
# Cell 21: Run the complete prediction process again for another patient.
symptoms = input("Enter the symptoms separated by comma: ")

symptoms = [s.strip() for s in symptoms.split(",")]

new_patient = pd.DataFrame(
    0,
    index=[0],
    columns=xtrain.columns
)

for symptom in symptoms:
    if symptom in new_patient.columns:
        new_patient[symptom] = 1
    else:
        print("Symptom not found:", symptom)

prediction = model.predict(new_patient)
disease = encoder.inverse_transform(prediction)

print("Predicted Disease:", disease[0])
```
