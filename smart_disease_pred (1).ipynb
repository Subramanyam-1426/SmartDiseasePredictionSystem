```python
# ============================================================
# SMART DISEASE PREDICTION SYSTEM
# RANDOM FOREST CLASSIFIER
# ============================================================

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
# 1. LOAD DATASET

df = pd.read_csv("Training.csv")

# Remove unnecessary index column if present
df = df.drop(columns=["Unnamed: 133"], errors="ignore")
print(df)
```

```python
print("First 5 rows")
print(df.head())
```

```python
print("Last 5 rows")
print(df.tail())
```

```python
print("Dataset Shape:", df.shape)
```

```python

# 2. SEPARATE FEATURES AND TARGET

X = df.drop("prognosis", axis=1)
y = df["prognosis"]

print("Number of Symptoms:", X.shape[1])
print("Number of Diseases:", y.nunique())
```

```python

# 3. ENCODE DISEASE NAMES

encoder = LabelEncoder()

y_encoded = encoder.fit_transform(y)
```

```python

# 4. SPLIT DATASET

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("Training Samples:", X_train.shape[0])
print("Testing Samples:", X_test.shape[0])
```

```python

# 5. CREATE RANDOM FOREST MODEL

model = RandomForestClassifier(
    n_estimators=300,
    max_features="sqrt",
    random_state=42,
    n_jobs=-1
)
```

```python

# 6. TRAIN MODEL

model.fit(X_train, y_train)

print("\nModel trained successfully!")
```

```python

# 7. TEST MODEL

y_pred = model.predict(X_test)
```

```python

# 8. EVALUATE MODEL
accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

print("\n========== MODEL PERFORMANCE ==========")

print("Accuracy  :", round(accuracy * 100, 2), "%")
print("Precision :", round(precision * 100, 2), "%")
print("Recall    :", round(recall * 100, 2), "%")
print("F1 Score  :", round(f1 * 100, 2), "%")
```

```python

# 9. CLASSIFICATION REPORT

print("\n========== CLASSIFICATION REPORT ==========")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=encoder.classes_,
        zero_division=0
    )
)
```

```python

# 10. CONFUSION MATRIX

cm = confusion_matrix(y_test, y_pred)

print("\n========== CONFUSION MATRIX ==========")
print(cm)
```

```python

# 11. NEW PATIENT TESTING

# Create a new patient with all symptoms initially 0

new_patient = pd.DataFrame(
    [[0] * len(X.columns)],
    columns=X.columns
)



# ENTER PATIENT SYMPTOMS 

# 1 = symptom present
# 0 = symptom absent

new_patient["itching"] = 1
new_patient["skin_rash"] = 1
new_patient["nodal_skin_eruptions"] = 1


# 12. PREDICT DISEASE

prediction = model.predict(new_patient)

predicted_disease = encoder.inverse_transform(prediction)[0]

print("\n========== NEW PATIENT RESULT ==========")

print("Predicted Disease:", predicted_disease)
```

```python

# 13. PREDICTION PROBABILITY

probabilities = model.predict_proba(new_patient)
confidence = probabilities.max() * 100
print("Prediction Confidence:", round(confidence, 2), "%")
```
